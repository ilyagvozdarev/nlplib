### LIBRA

https://huggingface.co/datasets/ai-forever/LIBRA

```python
def load_n_test_sample_of_each_dataset_name(dataset_name, count):
    datasets = {}
    for name in get_dataset_config_names(dataset_name):
        datasets[name] = load_dataset(dataset_name, name=name, split=f'test[:{count}]')
    return datasets
```

```python
LIBRA = "ai-forever/LIBRA"
```

```python
datasets = load_n_test_sample_of_each_dataset_name(LIBRA, 3)
```
