```python
from transformers import pipeline, AutoTokenizer, AutoModelForTokenClassification

tokenizer = AutoTokenizer.from_pretrained("xlm-roberta-large-finetuned-conll03-english")
model = AutoModelForTokenClassification.from_pretrained("xlm-roberta-large-finetuned-conll03-english")
classifier = pipeline("ner", model=model, tokenizer=tokenizer, aggregation_strategy='first')
# print(classifier(["Alya told Jasmine that Andrew Garfield could pay with cash..", "Alya told Jasmine that Andrew Garfield could pay with cash.."]))
print(classifier("Alya told Jasmine that Andrew Garfield could pay with cash.." * 20000))

print(tokenizer.convert_tokens_to_string(['ads', 'dsa']))
```
