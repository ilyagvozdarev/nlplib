````python
import matplotlib.pyplot as plt
import matplotlib.patches as patches
import numpy as np


prefix = 'В наших клиниках активно используют внутреннюю телемедицину. С помощью современных'


def plot_strategy(probabilities, tokens, k_chosen=None, text=None, savefile=None):

    k_chosen = k_chosen or len(tokens)

    colors = ['salmon'] * k_chosen + ['black'] * (len(probabilities) - k_chosen)

    # Параметры
    width = 0.5
    spacing = 1.5
    shift = 0.5
    line_width = 3
    ylim_top = 0.8

    plt.rcParams['figure.dpi'] = 500
    plt.rcParams['savefig.dpi'] = 500
    plt.rcParams['font.family'] = 'Verdana'



    fig, ax = plt.subplots(figsize=(25, 15))

    # Настройки
    ax.set_xlim(-0.5, len(tokens) * spacing - spacing/2 + shift)

    # Рисуем прямоугольники с немного отрицательным Y, чтобы нижняя граница была видна
    for i, prob in enumerate(probabilities):
        x = shift + i * spacing
        y = 0.005  # Немного ниже нуля
        
        # Создаем прямоугольник
        rect = patches.Rectangle((x - width/2, y), width, prob - y, 
                            linewidth=line_width, 
                            edgecolor=colors[i], 
                            facecolor='white',
                            joinstyle='bevel',
                            hatch='//' if i < k_chosen else None)
        
        # Добавляем прямоугольник на график
        ax.add_patch(rect)
        
        # Добавляем подпись токена
        ax.text(x, -0.02, tokens[i], ha='center', va='top', fontfamily='Verdana', style='italic', fontsize=28)
        
        # Добавляем значение вероятности
        ax.text(x, prob + 0.01, f'{prob:.2f}', ha='center', va='bottom', fontfamily='Verdana', style='italic', fontsize=36)


    # Альтернативный вариант - линия в самом левом краю
    # Добавляем вертикальную пунктирную линию слева с цифрой 1.0
    left_edge, right_edge = ax.get_xlim()[0], ax.get_xlim()[1]  # самая левая граница графика
    print('right_edge =', right_edge)

    # Рисуем пунктирную линию от 0 до 1.0
    ax.plot([left_edge + 0.24, left_edge + 0.2], [0, 0.98], linestyle='--', color='gray', linewidth=2, alpha=0.7)

    # Добавляем подпись "1.0" сверху линии
    ax.text(left_edge, ylim_top, '1.0', ha='left', va='bottom', 
            fontfamily='Verdana', style='italic', fontsize=36, 
            color='gray')

    # Добавляем подпись с параметрами
    ax.text(right_edge - 7.0, ylim_top - 0.2, text, ha='left', va='bottom', 
            fontfamily='Verdana', style='italic', fontsize=36, 
            color='black')


    # Настройки
    ax.set_ylim(-0.01, ylim_top)
    print('ylim = ', max(probabilities) + 0.05)
    
    # Убираем оси
    ax.set_xticks([])
    ax.set_yticks([])
    for spine in ax.spines.values():
        spine.set_visible(False)



    # plt.title('Распределение вероятностей токенов', fontfamily='Verdana', style='italic')
    plt.tight_layout()

    # Сохраняем с очень высоким качеством
    if savefile:
        plt.savefig(savefile, dpi=500, bbox_inches='tight', 
                    facecolor='white', edgecolor='none')

    plt.show()
````

````python
# softmax

logits = [6, 5, 4, 3, 1, 0.5, 0.3, 0.1]
# probabilities = [0.4, 0.25, 0.15, 0.15, 0.1, 0.06, 0.02, 0.01]



probs = np.exp(logits) / np.sum(np.exp(logits))
print(probs)

logits_new = np.log(probs)
probs = np.exp(logits_new) / np.sum(np.exp(logits_new))
print(probs)
````

````python
prefix = 'В наших клиниках активно используют внутреннюю телемедицину. С помощью современных'

probabilities = [0.4, 0.20, 0.10, 0.10, 0.08, 0.05, 0.04, 0.03]
assert sum(probabilities) == 1.0
logits = np.log(probabilities)
probs = np.exp(logits) / np.sum(np.exp(logits))
print(probabilities)
print(probs)

tokens = ['технологий', 'систем', 'методов', 'решений', 'программ', 'подходов', 'инструментов', 'устройств']

plot_strategy(probs, tokens, k_chosen=1, text='жадное декодирование', savefile='жадное_декодирование.png')
plot_strategy(probs, tokens, text='random sampling', savefile='random_sampling.png')

temperature = 2
print(temperature)
probs_new = np.exp(logits/temperature) / np.sum(np.exp(logits/temperature))
print(probs_new)

plot_strategy(probs_new, tokens, text='random sampling\n\ntemperature = 2', savefile='random_sampling_temp2.png')

temperature = 0.7
print(temperature)
probs_new = np.exp(logits/temperature) / np.sum(np.exp(logits/temperature))
print(probs_new)

plot_strategy(probs_new, tokens, text='random sampling\n\ntemperature = 0.7', savefile='random_sampling_temp07.png')
````

````python
# top-p = 0.8

prefix = 'Для обработки данных в Python чаще всего используется библиотека.'

tokens = ['Pandas', 'NumPy', 'Matplotlib', 'sklearn', 'PyTorch', 'requests', 'Flask', 'Django', 'jinja2', 'json']

probs = np.array([0.43, 0.22, 0.18, 0.04, 0.04, 0.03, 0.02, 0.02 , 0.01, 0.01])

top_p = 0.8
k_chosen = [i for i in range(len(probs)) if sum(probs[:i]) > top_p][0]

plot_strategy(probs, tokens, k_chosen=k_chosen, text='top-p = 0.8', savefile='top_p08_example1.png')
````
