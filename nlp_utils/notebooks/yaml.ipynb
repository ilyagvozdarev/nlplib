```python
import yaml
```

```python
# переопределяем Dumper чтобы настраивать indent

class IndentDumper(yaml.Dumper):
    def increase_indent(self, flow=False, indentless=False):
        return super(IndentDumper, self).increase_indent(flow, False)
```

```python
path = r'C:\Users\el1ja\Desktop\repo\pivotal\configs\models\train_config.json'

with open(path, 'r', encoding='utf-8') as file:
    data = yaml.safe_load(file)                     # Используйте safe_load для безопасности


with open(path.replace('json', 'yaml'), 'w') as file:
    # yaml.safe_dump(data, file, sort_keys=False)
    yaml.dump(data, file, sort_keys=False, allow_unicode=True, Dumper=IndentDumper, indent=4)

# s = yaml.safe_dump(data, sort_keys=False, allow_unicode=True)     # safe_dump не поддерживает Dumper
```
