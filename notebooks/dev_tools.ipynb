```python
import os

# запускать из корня репозитория, рядом с pyproject.toml
PACKAGE_ROOT = "../nlplib"  

SKIP_DIRS = {"__pycache__", ".git", ".ipynb_checkpoints"}

for dirpath, dirnames, filenames in os.walk(PACKAGE_ROOT):
    dirnames[:] = [d for d in dirnames if d not in SKIP_DIRS]

    init_path = os.path.join(dirpath, "__init__.py")
    if not os.path.exists(init_path):
        open(init_path, "w").close()
        print(f"created: {init_path}")
    else:
        print(f"exists:  {init_path}")
```

```python
from pathlib import Path

def print_files_tree(dir_path: Path, indent: int = 0, skip = None):
    entries = sorted(dir_path.iterdir(), key=lambda e: (e.is_file(), e.name))
    for entry in entries:
        if entry.name in skip:
            continue
        print("   " * indent + "|- " + entry.name)
        if entry.is_dir():
            print_files_tree(entry, indent + 1, skip)

p = Path('../nlplib')
print_files_tree(p, skip=['__init__.py'])
```
